```python
import pandas as pd
import glob

print("--- 1. LENDO E UNIFICANDO OS ARQUIVOS CSV ---")
# Procura todos os arquivos mensais na pasta atual do Colab
arquivos_csv = sorted(glob.glob('dados_dengue_mes*.csv'))
print(f"Arquivos encontrados: {arquivos_csv}")

if len(arquivos_csv) > 0:
    # Lê cada arquivo usando o separador ';' correto e unifica
    lista_dfs = [pd.read_csv(arq, sep=';', encoding='utf-8') for arq in arquivos_csv]
    df_total = pd.concat(lista_dfs, ignore_index=True)
    print(f"Total de linhas unificadas: {len(df_total)}")

    print("\n--- 2. APLICANDO O TRATAMENTO ÉTICO E ENGENHARIA DE DADOS ---")
    df_tratado = df_total.copy()

    # Padroniza os nomes das colunas
    df_tratado.columns = [col.strip().lower() for col in df_tratado.columns]

    # Tratamento Temporal (Forward Fill nas datas)
    df_tratado['data'] = pd.to_datetime(df_tratado['data'], format='%d/%m/%Y', errors='coerce')
    df_tratado['data'] = df_tratado['data'].ffill()

    # Propagação correta dos bairros (local/atividade)
    if 'local/atividade' in df_tratado.columns:
        df_tratado['local/atividade'] = df_tratado['local/atividade'].ffill()

    # Tratamento Numérico (vazios viram 0)
    cols_numericas = ['qtd', 'focos eliminados', 'código']
    for col in cols_numericas:
        if col in df_tratado.columns:
            df_tratado[col] = pd.to_numeric(df_tratado[col], errors='coerce').fillna(0)

    # Tratamento de Textos
    for col in ['local/atividade', 'tipo de recipiente']:
        if col in df_tratado.columns:
            df_tratado[col] = df_tratado[col].fillna('não informado').astype(str).str.lower().str.strip()

    print("\n--- 3. TUDO PRONTO COM SUCESSO! ---")
    display(df_tratado.head(10))

else:
    print("\n[ATENÇÃO]: Nenhum arquivo 'dados_dengue_mes*.csv' foi encontrado na pasta do Colab.")
    print("Por favor, arraste os arquivos CSV para o ícone de pasta na barra lateral esquerda antes de rodar novamente.")
```

```python
import pandas as pd
import glob
from google.colab import drive

# 1. Monta o Google Drive para salvamento seguro
drive.mount('/content/drive')

# 2. Busca todos os arquivos mensais na pasta /content
arquivos_csv = sorted(glob.glob('/content/dados_dengue_mes*.csv'))
print("Arquivos encontrados:", arquivos_csv)

if len(arquivos_csv) > 0:
    # 3. Lê e unifica todos os arquivos usando sep=';'
    lista_dfs = [pd.read_csv(arq, sep=';', encoding='utf-8') for arq in arquivos_csv]
    df_total = pd.concat(lista_dfs, ignore_index=True)

    print(f"\nTotal de linhas unificadas de todos os meses: {len(df_total)}")

    # 4. Cópia de trabalho para a etapa de tratamento (Engenharia de Dados)
    df_tratado = df_total.copy()

    # Padroniza os nomes das colunas para letras minúsculas para evitar erros de digitação
    df_tratado.columns = [col.strip().lower() for col in df_tratado.columns]

    # 5. Tratamento Temporal: Converte data e propaga para as linhas do mesmo dia (Forward Fill)
    df_tratado['data'] = pd.to_datetime(df_tratado['data'], format='%d/%m/%Y', errors='coerce')
    df_tratado['data'] = df_tratado['data'].ffill()

    # 6. Propagação correta do local/bairro (linhas secundárias herdam o local de cima)
    df_tratado['local/atividade'] = df_tratado['local/atividade'].ffill()

    # 7. Tratamento honesto de numéricos (vazios viram 0)
    cols_numericas = ['qtd', 'focos eliminados', 'código']
    for col in cols_numericas:
        if col in df_tratado.columns:
            df_tratado[col] = pd.to_numeric(df_tratado[col], errors='coerce').fillna(0)

    # 8. Tratamento de textos restantes
    for col in ['local/atividade', 'tipo de recipiente']:
        if col in df_tratado.columns:
            df_tratado[col] = df_tratado[col].fillna('não informado').astype(str).str.lower().str.strip()

    # 9. Salva o resultado tratado diretamente no Google Drive
    caminho_drive = '/content/drive/MyDrive/dados_dengue_tratados_finais.csv'
    df_tratado.to_csv(caminho_drive, index=False)
    print(f"\nSucesso absoluto! Base tratada salva com segurança em: {caminho_drive}")

    # Mostra uma amostra do resultado final
    display(df_tratado.head(15))
else:
    print("[ERRO]: Nenhum arquivo CSV foi encontrado na pasta /content.")
```
